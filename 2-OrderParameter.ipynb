# 2.1. Order Parameter 

```python
import numpy as np
import pandas as pd
import os

from aux_functions import order_parameter
```

## High coupling

```python
for typenoise in ["OutputTimeDep", "OutputColumnar"]:
    
    for delta, deltaname in zip([0, np.arctan(5)], ["0", "atan5"]):

        path = f"{typenoise}/Solutions_delta{deltaname}.pkl"
        if not os.path.exists(path):
            print(f"File {path} not found, skipping.")
            continue

        print(f"## {typenoise} – Delta {deltaname}")

        # Load previously saved simulations
        df = pd.read_pickle(path)

        R_list = []
        Psi_list = []

        # Loop over simulations (one per seed / row)
        for _, row in df.iterrows():
            phases = row["phases"]

            # ensure array, not list-of-lists
            R, Psi = order_parameter(phases)

            R_list.append(R)
            Psi_list.append(Psi)

        # Avoid chained-assignment issues
        df = df.copy()

        # assign one array per row, positionally
        df["R"] = R_list
        df["Psi"] = Psi_list

        # Save back the updated simulations DataFrame (with roughness column)
        df.to_pickle(path)

        print("Seeds (index):", df.index.tolist())
        print("Columns:", df.columns.tolist())
```
