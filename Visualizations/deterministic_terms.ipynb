# Visualization of the evolution of the deterministic terms

```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(['science', 'no-latex'])
```

## Surface tension (diffusion)

```python
# Setup domain
x = np.linspace(-3, 3, 500)
dx = x[1] - x[0]

# (a) Initial Gaussian surface h(x, t)
h = np.exp(-x**2)

# (b) Surface Tension term: nu * grad^2(h)
# Using second-order central difference for Laplacian
laplacian = np.gradient(np.gradient(h, dx), dx)
nu = 0.2
term_b = nu * laplacian

# (c) Evolution: h(x, t + dt)
dt = 0.5
h_new = h + term_b * dt

# Plotting
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8), sharex=True)

ax1.plot(x, h, 'k-')
ax1.set_ylabel('$h(x, t)$')
ax1.text(0.02, 0.8, '(a)', transform=ax1.transAxes, fontsize=12, fontweight='bold')

ax2.plot(x, term_b, 'k-')
# ax2.axhline(0, color='black', lw=0.5)
ax2.set_ylabel(r'$\nu \nabla^2 h$')
ax2.text(0.02, 0.8, '(b)', transform=ax2.transAxes, fontsize=12, fontweight='bold')

ax2.set_ylim(-0.5,0.3)

ax3.plot(x, h, 'k--', label='original')
ax3.plot(x, h_new, 'k-')
ax3.set_ylabel('$h(x, t + \delta t)$')
ax3.set_xlabel('$x$')
ax3.text(0.02, 0.8, '(c)', transform=ax3.transAxes, fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

## Lateral growth

```python
# Setup domain
x = np.linspace(-3, 3, 500)
dx = x[1] - x[0]

# (a) Initial Gaussian surface h(x, t)
h = np.exp(-x**2)

# (b) Lateral Growth term: lambda * (grad h)^2
grad_h = np.gradient(h, dx)
lmbda = 0.2 # coefficient
term_b = lmbda * (grad_h**2)

# (c) Evolution: h(x, t + dt)
dt = 0.5
h_new = h + term_b * dt

# Plotting
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8), sharex=True)

ax1.plot(x, h, 'k-')
ax1.set_ylabel('$h(x, t)$')
ax1.text(0.02, 0.8, '(a)', transform=ax1.transAxes, fontsize=12, fontweight='bold')

ax2.plot(x, term_b, 'k-')
ax2.set_ylabel(r'$\lambda (\nabla h)^2$')
ax2.text(0.02, 0.8, '(b)', transform=ax2.transAxes, fontsize=12, fontweight='bold')

ax2.set_ylim(-0.05,0.25)

ax3.plot(x, h, 'k--', label='original')
ax3.plot(x, h_new, 'k-')
ax3.set_ylabel('$h(x, t + \delta t)$')
ax3.set_xlabel('$x$')
ax3.text(0.02, 0.8, '(c)', transform=ax3.transAxes, fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
